```julia
using JLD2
using Plots
using LaTeXStrings
using LinearAlgebra
using Polynomials
using Printf
```

```julia
# ------------------------------------------------------------
# Load data
# ------------------------------------------------------------
@load "./data/gamma_scan_2Delta3.jld2" gamma_list M_kernel_matrix_list

# ------------------------------------------------------------
# Extract the spectral norm of the kappa matrices
# ------------------------------------------------------------
kappa_2Delta3 = [0.5 * gamma_list[i]^2 * opnorm(M_kernel_matrix_list[i]) for i in eachindex(gamma_list)]

p1 = plot(
    gamma_list, kappa_2Delta3,
    marker = :circle,
    xscale = :log10,
    yscale = :log10,
    xlabel = L"\gamma",
    title = L"\mathrm{Double-}\Delta_3",
    label = "numerical result",
    legend = :bottomright
)

# top-left formula (like your current style)
annotate!(
    (0.00001, 1.02),
    text(L"\frac{\gamma^2}{2}\,\|\kappa \|", 12)
);

p = fit(log10.(gamma_list[end-20:end]), log10.(kappa_2Delta3[end-20:end]), 1) 
a, b = coeffs(p)
fit_curve = 10 .^ (a .+ b .* log10.(gamma_list))
# ------------------------------------------------------------
# Show plots
# ------------------------------------------------------------
plot!(
    p1,
    gamma_list,
    fit_curve,
    lw = 2,
    linestyle = :dash,
    label = "power-law fit (slope = $(round(b, digits=6)))"
)

display(p1)
```
